```python
import time
x=time.time()
def work(x):
    time.sleep(3)
    return x * x
for i in range(4):

    print(work(i))
print(time.time()-x)
```

```python
from concurrent.futures import ThreadPoolExecutor
# from concurrent.futures import ProcessPoolExecutor
x=time.time()
def work(x):
    time.sleep(3)
    return x * x
with ThreadPoolExecutor(max_workers=4) as executor:
    results = executor.map(work, range(4))

print(list(results))
print(time.time()-x)
```
